# Cálculo de Payoff das opções

```python
import numpy as np
import matplotlib.pyplot as plt
```

1. Call Comprada

```python
# 1. Definindo a função de call comprada

def payoff_call_comprada(s, strike_price, premio):
    return np.maximum(s - strike_price, 0) - premio
```

```python
# 2. Definindo o strike price (Preço de Exercício) e o premio (valor pago pela opção)

strike_price = 50
premio = 2
```

```python
# 3. Gerando valores para o preço do ativo subjacente (S)

s_values = np.linspace(0, 100, 1000)
```

```python
# 4. Calculando os payoffs da opção

payoffs = payoff_call_comprada(s_values, strike_price, premio)
```

```python
plt.figure(figsize= (10,6))
plt.plot(s_values, payoffs, label = 'Payoff da Call Comprada', color = 'blue')
plt.title('Payoff de uma Opção de Compra (Call) Comprada')
plt.xlabel('Preço do Ativo Subjacente (S)')
plt.ylabel('Payoff')
plt.axhline(0, color = 'black', linestyle = '--')
plt.axvline(strike_price, color = 'red', linestyle = '--', label = 'Preço de Exercício')
plt.legend()
plt.grid(True)
plt.show();
```

2. Call vendida

```python
def payoff_call_vendida(s, strike_price, premio):
    return -(np.maximum(s - strike_price, 0) - premio)

strike_price = 50
premio = 2

s_values = np.linspace(0, 100, 1000)

payoffs = payoff_call_vendida(s_values, strike_price, premio)
```

```python
plt.figure(figsize= (10,6))
plt.plot(s_values, payoffs, label = 'Payoff da Call Vendida', color = 'green')
plt.title('Payoff de uma Opção de Compra (Call) Vendida')
plt.xlabel('Preço do Ativo Subjacente (S)')
plt.ylabel('Payoff')
plt.axhline(0, color = 'black', linestyle = '--')
plt.axvline(strike_price, color = 'red', linestyle = '--', label = 'Preço de Exercício')
plt.legend()
plt.grid(True)
plt.show();
```

3. Put Comprada

```python
def payoff_put_comprada(s, strike_price, premio):
    return np.maximum(strike_price - s, 0) - premio

strike_price = 50
premio = 2

s_values = np.linspace(0, 100, 1000)

payoffs = payoff_put_comprada(s_values, strike_price, premio)
```

```python
plt.figure(figsize= (10,6))
plt.plot(s_values, payoffs, label = 'Payoff da Put Comprada', color = 'blue')
plt.title('Payoff de uma Opção de Venda (Put) Comprada')
plt.xlabel('Preço do Ativo Subjacente (S)')
plt.ylabel('Payoff')
plt.axhline(0, color = 'black', linestyle = '--')
plt.axvline(strike_price, color = 'red', linestyle = '--', label = 'Preço de Exercício')
plt.legend()
plt.grid(True)
plt.show();
```
