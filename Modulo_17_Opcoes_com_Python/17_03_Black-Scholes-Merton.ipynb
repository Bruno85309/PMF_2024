## Modelo de Black-Scholes-Merton

### Definindo a função

```python
import math

def black_scholes_call_put(S, K, T, r, sigma, option_type):
    """
    Calcula o preço de uma opção de compra (call) ou venda (put) usando o modelo de Black-Scholes.

    Parâmetros:
    S : Preço atual do ativo subjacente
    K : Preço de exercício da opção
    T : Tempo até o vencimento em anos
    r : Taxa de juros livre de risco
    sigma : Volatilidade do ativo subjacente
    option_type : Tipo de opção ('call' para opção de compra, 'put' para opção de venda)

    Retorna:
    Preço da opção calculado pelo modelo de Black-Scholes-Merton
    """
    # Calcula d1 e d2 conforme as equações de Black-Scholes-Merton
    d1 = (math.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * math.sqrt(T))
    d2 = d1 - sigma * math.sqrt(T)
    
    # Calcula o preço da opção de acordo com o tipo (call ou put)
    if option_type == 'call':
        option_price = S * norm_cdf(d1) - K * math.exp(-r * T) * norm_cdf(d2)
    elif option_type == 'put':
        option_price = K * math.exp(-r * T) * norm_cdf(-d2) - S * norm_cdf(-d1)
    else:
        raise ValueError("Tipo de opção inválido. Use 'call' ou 'put'.")
    
    return option_price

def norm_cdf(x):
    """
    Função de distribuição cumulativa normal padrão.
    
    Parâmetros:
    x : Valor para o qual calcular a CDF

    Retorna:
    Valor da CDF no ponto x
    """
    return (1.0 + math.erf(x / math.sqrt(2.0))) / 2.0
```

### Exemplo Aula

```python
##### Exemplo de uso #####

# Parâmetros para o exemplo
S = 34.63        # Preço spot do ativo subjacente
K = 34.31        # Preço de exercício
T = 23/252        # Vencimento como fração do ano, base 252
r = 0.1075     # Taxa de juros livre de risco
sigma = 0.22     # Volatilidade dos retornos do ativo subjacente
```

```python
# Calcula o preço da opção de compra (call)
call_price = black_scholes_call_put(S, K, T, r, sigma, 'call')

# Calcula o preço da opção de venda (put)
put_price = black_scholes_call_put(S, K, T, r, sigma, 'put')

# Exibe os resultados
print(f"Preço da opção de compra (call): {call_price:.2f}")
print(f"Preço da opção de venda (put): {put_price:.2f}")
```
