# Gregas (Greeks) em Opções

```python
# Estimação das Gregas

import numpy as np
from scipy.stats import norm

# Parametros - exemplo
S = 34.63          # Preço spot do ativo subjacente
K = 34.31           # Preço de exercício
T = 23/252          # Vencimento como fração do ano, base 252
r = 0.1075        # Taxa de juros livre de risco
sigma = 0.22        # Volatilidade dos retornos do ativo subjacente
```

## 1. Delta

O Delta (Δ) exprime exatamente a razão entre a mudança no preço da
opção e a mudança no preço da ação subjacente.

```python
### Delta
def black_scholes_delta(S, K, T, r, sigma, option_type='call'):
    d1 = (np.log(S / K) + (r + (sigma**2) / 2) * T) / (sigma * np.sqrt(T))
    
    if option_type == 'call':
        delta = norm.cdf(d1)
    elif option_type == 'put':
        delta = (norm.cdf(d1) - 1)
    else:
        raise ValueError("Tipo de opção deve ser 'call' ou 'put'.")
    
    return delta

# Calculando o Delta
delta_call = black_scholes_delta(S, K, T, r, sigma, option_type='call')
delta_put = black_scholes_delta(S, K, T, r, sigma, option_type='put')
```

## 2. Theta

O Theta (Θ) mede a mudança no preço da opção em relação à passagem
do tempo.

```python
# Calculando o Theta da opção de compra
def theta_call(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    
    theta = -(S * norm.pdf(d1) * sigma / (2 * np.sqrt(T))) - r * K * np.exp(-r * T) * norm.cdf(d2)
    
    return theta

theta_call = theta_call(S, K, T, r, sigma)   # base 252
theta_call 
```

```python
# Calculando o Theta da opção de venda
def theta_put(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    
    theta = -(S * norm.pdf(-d1) * sigma / (2 * np.sqrt(T))) + r * K * np.exp(-r * T) * norm.cdf(-d2)
    
    return theta

theta_put = theta_put(S, K, T, r, sigma)     # base 252
theta_put
```

## 3. Gamma

O Gamma (Γ) mede a taxa de mudança no Delta (Δ) a opção em relação a
mudanças no preço do ativo subjacente.

```python
def gamma(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    gamma = norm.pdf(d1) / (S * sigma * np.sqrt(T))
    return gamma

# Calculando o Gamma da opção
gamma = gamma(S, K, T, r, sigma)
gamma
```

## 4. Vega

O Vega (V) mede a taxa de mudança no preço da opção em relação a
mudanças na volatilidade.

```python
def vega(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    vega = S * norm.pdf(d1) * np.sqrt(T)
    return vega

# Calculando o Vega da opção
vega = vega(S, K, T, r, sigma)
vega
```

## 5. Rho
O Rho (ρ) mede a taxa de mudança no preço da opção em relação a
mudanças na taxa de juros livre de risco.

```python
# Calculando o Rho da opção de compra
def rho_call(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    
    rho = K * T * np.exp(-r * T) * norm.cdf(d2)
    return rho

rho_call = rho_call(S, K, T, r, sigma)
rho_call
```

```python
# Calculando o Rho da opção de venda
def rho_put(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    
    rho = -K * T * np.exp(-r * T) * norm.cdf(-d2)
    return rho

rho_put = rho_put(S, K, T, r, sigma)
rho_put
```
